## Question 12 — Electricity Bill Function

**Problem:**  
An electricity company charges different rates based on units consumed. Create a function that calculates the bill.

**Suggested rates (you may adjust slightly):**
- 0 – 100 units → Rs. 5 per unit
- 101 – 200 units → Rs. 10 per unit
- Above 200 units → Rs. 15 per unit

**Requirements:**
- Create a function that accepts the number of **units consumed**.
- Calculate and **return** the final electricity bill using conditions.
- Display the returned bill outside the function.

**Focus:** Function + arguments + conditions + `return`

````python
def calculate_bill(units):
    if units <= 100:
        bill = units * 5
    elif units <= 200:
        bill = (100 * 5) + (units - 100) * 10
    else:
        bill = (100 * 5) + (100 * 10) + (units - 200) * 15
    return bill


# Receive the returned bill and display it outside the function
bill1 = calculate_bill(80)
print("Units: 80  | Bill: Rs.", bill1)

bill2 = calculate_bill(150)
print("Units: 150 | Bill: Rs.", bill2)

bill3 = calculate_bill(250)
print("Units: 250 | Bill: Rs.", bill3)
````

## Question 16 — Student Marks Processor

**Problem:**  
A teacher has a list of marks of all students in a class and wants key statistics.

**Requirements:**
- Create a function that accepts a **list of student marks**.
- Using a loop (or appropriate logic), calculate and **return**:
  - Highest mark
  - Lowest mark
  - Total marks
  - Average marks
- Display the results outside the function.

**Focus:** Function + list + loop + `return`

````python
def process_marks(marks):
    highest = marks[0]
    lowest = marks[0]
    total = 0

    for mark in marks:
        total += mark
        if mark > highest:
            highest = mark
        if mark < lowest:
            lowest = mark

    average = total / len(marks)
    return highest, lowest, total, average


# Sample class marks
class_marks = [78, 92, 65, 88, 45, 70, 95, 59]

# Receive the returned values outside the function
highest, lowest, total, average = process_marks(class_marks)

# Display them
print("Highest Mark  :", highest)
print("Lowest Mark   :", lowest)
print("Total Marks   :", total)
print(f"Average Marks : {average:.2f}")
````

## Question 23 — Nested Salary Calculator

**Problem:**  
Salary calculation has several smaller steps. Use nested functions to organize the logic cleanly.

**Requirements:**
- Create an outer function called `calculate_salary()`.
- Inside it, create separate **nested functions** for:
  - Calculating allowance
  - Calculating deduction
  - Calculating final salary
- The outer function should combine the results and return the final salary.
- Call the outer function and display the result.

**Focus:** Nested functions

````python
def calculate_salary(basic):

    # Nested function 1: allowance (20% of basic)
    def calculate_allowance():
        return basic * 20 / 100

    # Nested function 2: deduction (10% of basic)
    def calculate_deduction():
        return basic * 10 / 100

    # Nested function 3: final salary
    def calculate_final(allowance, deduction):
        return basic + allowance - deduction

    # Outer function combines the results
    allowance = calculate_allowance()
    deduction = calculate_deduction()
    final_salary = calculate_final(allowance, deduction)

    return final_salary


# Call the outer function and display the result
salary = calculate_salary(50000)
print(f"Final Salary: Rs. {salary:,.2f}")
````

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
def process_order(price, quantity, discount_percent, tax_percent):

    # Nested function 1: subtotal
    def calculate_subtotal():
        return price * quantity

    # Nested function 2: discount amount
    def calculate_discount(subtotal):
        return subtotal * discount_percent / 100

    # Nested function 3: tax amount (charged on the discounted amount)
    def calculate_tax(amount):
        return amount * tax_percent / 100

    # Nested function 4: final amount
    def calculate_final(discounted_amount, tax):
        return discounted_amount + tax

    # Outer function combines the steps
    subtotal = calculate_subtotal()
    discount = calculate_discount(subtotal)
    discounted_amount = subtotal - discount
    tax = calculate_tax(discounted_amount)
    final_amount = calculate_final(discounted_amount, tax)

    return final_amount


# Demonstration with sample values
order1 = process_order(1500, 4, 10, 16)
print(f"Order 1 Final Amount: Rs. {order1:,.2f}")

order2 = process_order(800, 3, 5, 8)
print(f"Order 2 Final Amount: Rs. {order2:,.2f}")
````

## Question 26 — Lambda With Student Marks

**Problem:**  
You have a list of student marks. You need to process them quickly using lambda.

**Requirements:**
- Given a list of student marks, use a **lambda** function (with `map` or list comprehension) to create a new list containing the **square** of each mark.
- Create another lambda-based operation that checks whether a mark is **even** or not.
- Display both results clearly.

**Focus:** Lambda + list processing

````python
marks = [78, 92, 65, 88, 45, 70]

# Lambda 1: square of a mark
square = lambda m: m ** 2

# Lambda 2: check whether a mark is even
is_even = lambda m: m % 2 == 0

# Using map() with the square lambda
squared_marks = list(map(square, marks))

# Using map() with the even-check lambda
even_check = list(map(is_even, marks))

print("Original Marks :", marks)
print("Squared Marks  :", squared_marks)
print("Even Check     :", even_check)

print()
print("Mark-by-mark result:")
for mark, result in zip(marks, even_check):
    status = "Even" if result else "Odd"
    print(f"{mark} -> {status}")

````

````python
def get_employee_info(emp_id, name, department, designation):
    """Employee information: returns the details as a dictionary."""
    return {
        "id": emp_id,
        "name": name,
        "department": department,
        "designation": designation
    }


def get_basic_salary(basic):
    """Basic salary: rejects invalid values."""
    if basic <= 0:
        return 0
    return basic


def calculate_allowance(basic):
    """Allowance: 40% of basic salary."""
    return basic * 40 / 100


def calculate_deduction(basic):
    """Deduction: 5% provident fund on basic salary."""
    return basic * 5 / 100


def calculate_tax(gross):
    """Tax: monthly slabs applied on gross salary."""
    if gross <= 50000:
        return 0
    elif gross <= 100000:
        return (gross - 50000) * 5 / 100
    elif gross <= 200000:
        return 2500 + (gross - 100000) * 10 / 100
    else:
        return 12500 + (gross - 200000) * 15 / 100


def calculate_final_salary(basic, allowance, deduction, tax):
    """Final salary: basic + allowance - deduction - tax."""
    return basic + allowance - deduction - tax


def display_report(info, basic, allowance, deduction, tax, final):
    """Prints a clean salary report."""
    print("=" * 50)
    print("EMPLOYEE SALARY REPORT".center(50))
    print("=" * 50)
    print(f"Employee ID   : {info['id']}")
    print(f"Name          : {info['name']}")
    print(f"Department    : {info['department']}")
    print(f"Designation   : {info['designation']}")
    print("-" * 50)
    print(f"Basic Salary  : Rs. {basic:>12,.2f}")
    print(f"Allowance     : Rs. {allowance:>12,.2f}")
    print(f"Deduction     : Rs. {deduction:>12,.2f}")
    print(f"Tax           : Rs. {tax:>12,.2f}")
    print("-" * 50)
    print(f"Final Salary  : Rs. {final:>12,.2f}")
    print("=" * 50)


def process_employee(emp_id, name, department, designation, basic_salary):
    """Connects all the functions for one employee."""
    info = get_employee_info(emp_id, name, department, designation)
    basic = get_basic_salary(basic_salary)
    allowance = calculate_allowance(basic)
    deduction = calculate_deduction(basic)
    gross = basic + allowance
    tax = calculate_tax(gross)
    final = calculate_final_salary(basic, allowance, deduction, tax)
    display_report(info, basic, allowance, deduction, tax, final)


# Run the payroll system
process_employee("E-101", "Ayesha Khan", "Accounts", "Senior Accountant", 80000)
print()
process_employee("E-102", "Ali Raza", "IT", "Developer", 40000)
````

## Question 28 — Smart Shopping Billing System

**Problem:**  
Build a shopping billing system that can handle multiple products.

**Requirements:**
- Allow the user (or hard-code) multiple products.
- Calculate:
  - Individual product totals
  - Overall subtotal
  - Discount
  - Tax
  - Final bill
- Use **multiple functions** to separate responsibilities.
- At least one function must use `*args` or a flexible-argument design.
- Display a professional-looking bill.

**Focus:** Real-world application + multiple functions + logic building

````python
def calculate_item_total(price, quantity):
    """Individual product total."""
    return price * quantity


def calculate_subtotal(*item_totals):
    """Overall subtotal: accepts any number of item totals (*args)."""
    return sum(item_totals)


def calculate_discount(subtotal):
    """Discount slabs based on subtotal. Returns (discount_amount, percent)."""
    if subtotal >= 10000:
        percent = 10
    elif subtotal >= 5000:
        percent = 5
    else:
        percent = 0
    return subtotal * percent / 100, percent


def calculate_tax(amount, tax_percent):
    """Tax on the discounted amount."""
    return amount * tax_percent / 100


def calculate_final_bill(discounted_amount, tax):
    """Final bill: discounted amount + tax."""
    return discounted_amount + tax


def display_bill(products, subtotal, discount, discount_percent,
                 tax, tax_percent, final_bill):
    """Prints a professional-looking bill."""
    print("=" * 48)
    print("SMART SHOPPING BILL".center(48))
    print("=" * 48)
    print(f"{'Item':<18}{'Price':>10}{'Qty':>6}{'Total':>14}")
    print("-" * 48)

    for name, price, quantity in products:
        total = calculate_item_total(price, quantity)
        print(f"{name:<18}{price:>10,.2f}{quantity:>6}{total:>14,.2f}")

    print("-" * 48)
    print(f"{'Subtotal':<34}{subtotal:>14,.2f}")
    print(f"{f'Discount ({discount_percent}%)':<34}{discount:>14,.2f}")
    print(f"{f'Tax ({tax_percent}%)':<34}{tax:>14,.2f}")
    print("=" * 48)
    print(f"{'FINAL BILL':<34}{final_bill:>14,.2f}")
    print("=" * 48)


def generate_bill(products, tax_percent=16):
    """Connects all the functions for one shopping cart."""
    item_totals = [calculate_item_total(price, qty) for _, price, qty in products]
    subtotal = calculate_subtotal(*item_totals)          # unpack list into *args
    discount, discount_percent = calculate_discount(subtotal)
    discounted_amount = subtotal - discount
    tax = calculate_tax(discounted_amount, tax_percent)
    final_bill = calculate_final_bill(discounted_amount, tax)
    display_bill(products, subtotal, discount, discount_percent,
                 tax, tax_percent, final_bill)


# Cart 1: large order (qualifies for 10% discount)
cart1 = [
    ("Keyboard", 2500, 2),
    ("Mouse", 800, 3),
    ("USB Cable", 350, 5),
    ("Headphones", 4200, 1),
]
generate_bill(cart1)
print()

# Cart 2: small order (no discount)
cart2 = [
    ("Notebook", 150, 4),
    ("Pen", 30, 10),
]
generate_bill(cart2)

````

## Question 29 — Student Result Management System

**Problem:**  
Create a complete student result system using functions.

**The system should:**
1. Accept student information
2. Accept multiple subject marks
3. Calculate total
4. Calculate percentage
5. Determine grade
6. Determine pass/fail
7. Display a final result card

**Requirements:**
- Use **multiple functions**
- Use arguments and `return`
- Use lists
- Use conditions and loops
- Use at least one advanced concept (`*args`, `**kwargs`, nested functions, or lambda)
- Do **not** write the entire program inside one function

**Focus:** Combining multiple function concepts

````python
def get_student_info(**details):
    """Student information: accepts any details using **kwargs."""
    return details


def collect_marks(*marks):
    """Accepts any number of subject marks using *args and returns a list."""
    return list(marks)


# Lambda: checks whether a single subject mark is passing (40 or above)
subject_passed = lambda mark: mark >= 40


def calculate_total(marks):
    """Total marks, using a loop."""
    total = 0
    for mark in marks:
        total += mark
    return total


def calculate_percentage(total, max_total):
    """Percentage from total and maximum possible marks."""
    return total / max_total * 100


def determine_grade(percentage):
    """Grade from percentage, using conditions."""
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    elif percentage >= 40:
        return "E"
    else:
        return "F"


def determine_result(percentage, marks):
    """Pass only if percentage >= 40 AND every subject is passed."""
    failed_subjects = [m for m in marks if not subject_passed(m)]
    if percentage >= 40 and len(failed_subjects) == 0:
        return "PASS"
    return "FAIL"


def display_result_card(info, subjects, marks, total, max_total,
                        percentage, grade, result):
    """Prints the final result card."""
    print("=" * 50)
    print("STUDENT RESULT CARD".center(50))
    print("=" * 50)
    print(f"{'Name':<12}: {info['name']}")
    print(f"{'Roll No':<12}: {info['roll_no']}")
    print(f"{'Course':<12}: {info['course']}")
    print(f"{'Institute':<12}: {info['institute']}")
    print("-" * 50)
    print(f"{'Subject':<24}{'Marks':>12}{'Remark':>14}")
    print("-" * 50)

    for subject, mark in zip(subjects, marks):
        remark = "Pass" if subject_passed(mark) else "Fail"
        print(f"{subject:<24}{mark:>12}{remark:>14}")

    print("-" * 50)
    print(f"{'Total Marks':<12}: {total} / {max_total}")
    print(f"{'Percentage':<12}: {percentage:.2f}%")
    print(f"{'Grade':<12}: {grade}")
    print(f"{'Result':<12}: {result}")
    print("=" * 50)


def generate_result(info, subjects, marks):
    """Connects all the functions for one student."""
    total = calculate_total(marks)
    max_total = len(marks) * 100
    percentage = calculate_percentage(total, max_total)
    grade = determine_grade(percentage)
    result = determine_result(percentage, marks)
    display_result_card(info, subjects, marks, total, max_total,
                        percentage, grade, result)


subjects = ["English", "Mathematics", "Physics", "Chemistry", "Computer Science"]

# Student 1
info1 = get_student_info(name="Ayesha Khan", roll_no="S-101",
                         course="Python Programming",
                         institute="Karachi Institute of Technology")
marks1 = collect_marks(78, 85, 72, 90, 88)
generate_result(info1, subjects, marks1)
print()

# Student 2
info2 = get_student_info(name="Ali Raza", roll_no="S-102",
                         course="Python Programming",
                         institute="Karachi Institute of Technology")
marks2 = collect_marks(45, 38, 30, 40, 35)
generate_result(info2, subjects, marks2)
````

## Question 30 — FINAL CHALLENGE: Mini E-Commerce Order System

**Problem:**  
Build a small console-based e-commerce order system using Python functions.

The system should simulate a customer placing an order and produce a professional-looking order summary.

**Suggested functions (you decide the exact design):**
- Customer information
- Product selection / product details
- Price & quantity calculation
- Subtotal
- Discount
- Tax
- Delivery charges
- Final bill
- Order summary display

**Mandatory Requirements:**
- Use multiple `def` functions

- Use conditions and loops
- Use lists and/or dictionaries
- Use `*args` **or** `**kwargs`
- Use **at least one nested function**
- Use **at least one lambda** function where it makes sense
- Keep different tasks in separate functions
- Do **not** write the entire program in one function

**Challenge:**  
Design the program yourself. Decide:
- What functions you need
- What arguments each function needs
- What each function should return
- Where conditions, loops, `*args`/`**kwargs`, nested functions, and lambda are useful

This question tests your **complete understanding** of Python functions.

**Focus:** Complete modular real-world system

````python


# Mini E-Commerce Order System

products = {
    1: {"name": "Laptop", "price": 120000},
    2: {"name": "Mouse", "price": 1500},
    3: {"name": "Keyboard", "price": 3500},
    4: {"name": "Headphones", "price": 5000},
    5: {"name": "Monitor", "price": 25000}
}


# 1. Customer Information
def customer_info(name, email, address):
    return {
        "name": name,
        "email": email,
        "address": address
    }


# 2. Display Products
def show_products(products):
    print("\n========== AVAILABLE PRODUCTS ==========")

    for key, value in products.items():
        print(f"{key}. {value['name']} - Rs. {value['price']:,}")


# 3. Add Product to Cart
def add_to_cart(cart, product_id, quantity):

    if product_id not in products:
        print("Invalid product!")
        return cart

    if quantity <= 0:
        print("Invalid quantity!")
        return cart

    product = products[product_id]

    cart.append({
        "name": product["name"],
        "price": product["price"],
        "quantity": quantity,
        "total": product["price"] * quantity
    })

    return cart


# 4. Calculate Subtotal
def calculate_subtotal(cart):
    subtotal = sum(item["total"] for item in cart)
    return subtotal


# 5. Calculate Discount
def calculate_discount(subtotal, discount_rate=0.10):

    if subtotal >= 100000:
        discount = subtotal * discount_rate

    elif subtotal >= 50000:
        discount = subtotal * 0.05

    else:
        discount = 0

    return discount


# 6. Calculate Tax
def calculate_tax(amount, tax_rate=0.05):
    return amount * tax_rate


# 7. Delivery Charges
def delivery_charges(subtotal, charge=250):

    if subtotal >= 10000:
        return 0

    return charge


# 8. Calculate Final Bill
def calculate_bill(subtotal, *args, **kwargs):

    discount = kwargs.get("discount", 0)
    tax = kwargs.get("tax", 0)
    delivery = kwargs.get("delivery", 0)

    final_bill = subtotal - discount + tax + delivery

    return final_bill


# 9. Order Summary
def order_summary(customer, cart, subtotal, discount, tax, delivery, final_bill):

    # Nested Function
    def print_line():
        print("=" * 50)

    # Lambda Function
    format_price = lambda price: f"Rs. {price:,.2f}"

    print_line()
    print("          E-COMMERCE ORDER SUMMARY")
    print_line()

    print(f"Customer Name : {customer['name']}")
    print(f"Email         : {customer['email']}")
    print(f"Address       : {customer['address']}")

    print_line()
    print("ORDER DETAILS")

    for item in cart:
        print(f"\nProduct  : {item['name']}")
        print(f"Price    : {format_price(item['price'])}")
        print(f"Quantity : {item['quantity']}")
        print(f"Total    : {format_price(item['total'])}")

    print_line()

    print(f"Subtotal         : {format_price(subtotal)}")
    print(f"Discount         : -{format_price(discount)}")
    print(f"Tax              : {format_price(tax)}")
    print(f"Delivery Charges : {format_price(delivery)}")

    print_line()

    print(f"FINAL BILL       : {format_price(final_bill)}")

    print_line()
    print("       THANK YOU FOR SHOPPING!")
    print_line()


# 10. Main Function
def main(customer, cart):

    if not cart:
        print("Your cart is empty!")
        return

    subtotal = calculate_subtotal(cart)

    discount = calculate_discount(subtotal)

    tax = calculate_tax(subtotal - discount)

    delivery = delivery_charges(subtotal)

    final_bill = calculate_bill(
        subtotal,
        discount=discount,
        tax=tax,
        delivery=delivery
    )

    order_summary(
        customer,
        cart,
        subtotal,
        discount,
        tax,
        delivery,
        final_bill
    )


# Program Execution

customer = customer_info(
    "Ahmer",
    "ahmer@gmail.com",
    "Karachi, Pakistan"
)

cart = []

show_products(products)

add_to_cart(cart, 1, 1)
add_to_cart(cart, 2, 2)
add_to_cart(cart, 3, 1)

main(customer, cart)
````
