## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].
    
    Ask user to enter marks for 5 students.

## 🏨 8. Hotel Room Booking

### Goal:
    Manage simple room booking.

### Task:

    Create a list of rooms [101, 102, 103, 104, 105]
    
    Ask user to choose a room.
    
    If room already booked, print “Room not available.”
    
    Continue until all rooms are booked.

### Example:

    Available rooms: [101, 102, 103, 104, 105]
    Choose room: 103
    Room 103 booked successfully!

```python
rooms = [101, 102, 103, 104, 105]
booked = []

while len(booked) < len(rooms):
    print("Available rooms:", [r for r in rooms if r not in booked])
    choice = int(input("Choose room: "))

    if choice not in rooms:
        print("Invalid room number")
    elif choice in booked:
        print("Room not available.")
    else:
        booked.append(choice)
        print(f"Room {choice} booked successfully!")

print("All rooms are booked!")
```

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
name = ""
balance = 0
account_created = False

while True:
    print("\n===== BANK MENU =====")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice == "1":
        if account_created:
            print("Account already exists.")
        else:
            name = input("Enter your name: ")
            balance = int(input("Enter opening balance: "))
            account_created = True
            print(f"Account created for {name}!")

    elif choice == "2":
        if not account_created:
            print("Please create an account first.")
        else:
            amount = int(input("Enter deposit amount: "))
            if amount <= 0:
                print("Invalid amount.")
            else:
                balance += amount
                print(f"{amount} deposited. New balance: {balance}")

    elif choice == "3":
        if not account_created:
            print("Please create an account first.")
        else:
            amount = int(input("Enter withdraw amount: "))
            if amount <= 0:
                print("Invalid amount.")
            elif amount > balance:
                print("Insufficient balance.")
            else:
                balance -= amount
                print(f"{amount} withdrawn. New balance: {balance}")

    elif choice == "4":
        if not account_created:
            print("Please create an account first.")
        else:
            print(f"Account holder: {name}")
            print(f"Balance: {balance}")

    elif choice == "5":
        print("Thank you for using our bank. Goodbye!")
        break

    else:
        print("Invalid choice. Try again.")
```

## School Marks Report
### Goal:

    Take input for multiple students’ marks in 3 subjects and calculate their total and average.

### Steps:

    Outer loop → for each student (3 students).
    
    Inner loop → take marks for 3 subjects.
    
    Store each student’s total in a list.
    
    Print all totals and class average.

### Example:
    Enter marks for Student 1:
    Subject 1: 80
    Subject 2: 75
    Subject 3: 90
    Enter marks for Student 2:
    Subject 1: 60
    Subject 2: 70
    Subject 3: 65
    → Output: [245, 195]

```python
totals = []

for student in range(1, 4):
    print(f"Enter marks for Student {student}:")
    total = 0

    for subject in range(1, 4):
        marks = int(input(f"Subject {subject}: "))
        total += marks

    totals.append(total)
    print(f"Student {student} average: {total / 3:.2f}")

print("Totals:", totals)
print("Class average:", sum(totals) / len(totals))
```

## Create a quiz game using loops and conditions.

```python
questions = [
    {"q": "Pakistan ki capital kya hai?", "options": ["Karachi", "Islamabad", "Lahore", "Quetta"], "answer": 2},
    {"q": "Python me list ka symbol kya hai?", "options": ["()", "{}", "[]", "<>"], "answer": 3},
    {"q": "2 + 3 * 4 = ?", "options": ["20", "14", "24", "12"], "answer": 2},
    {"q": "Quaid-e-Azam ka asli naam kya tha?", "options": ["Muhammad Ali Jinnah", "Liaquat Ali", "Allama Iqbal", "Sir Syed"], "answer": 1},
    {"q": "Python me loop ko rokne ke liye kya use hota hai?", "options": ["stop", "exit", "break", "end"], "answer": 3},
]

score = 0

print("===== QUIZ GAME =====")

for number, item in enumerate(questions, 1):
    print(f"\nQuestion {number}: {item['q']}")

    for i, option in enumerate(item["options"], 1):
        print(f"{i}. {option}")

    while True:
        try:
            choice = int(input("Your answer (1-4): "))
            if 1 <= choice <= 4:
                break
            print("1 se 4 ke beech number likho.")
        except ValueError:
            print("Sirf number likho.")

    if choice == item["answer"]:
        print("Correct!")
        score += 1
    else:
        print("Wrong! Sahi jawab:", item["options"][item["answer"] - 1])

print("\n===== RESULT =====")
print(f"Score: {score}/{len(questions)}")

percentage = score / len(questions) * 100

if percentage == 100:
    print("Excellent! Perfect score!")
elif percentage >= 60:
    print("Good job!")
elif percentage >= 40:
    print("Average, thodi aur mehnat karo.")
else:
    print("Better luck next time.")
```

    Ask if seat is booked (Y/N).

## Hotel Room Booking Tracker
### Goal:

    Record bookings for rooms on each floor.

### Steps:

    Outer loop → for 3 floors.
    
    Inner loop → for 4 rooms per floor.
    
    Ask user if booked (Y/N).
    
    Count total booked and available rooms.
    
    Print summary per floor.

### Example:
    Floor 1:
    Room 1: Y
    Room 2: N
    Room 3: N
    Room 4: Y
    Floor 2:
    Room 1: Y
    Room 2: Y
    Room 3: Y
    Room 4: Y
    → Output: Floor 1 = 2 booked, Floor 2 = 4 booked

```python
total_booked = 0
total_available = 0

for floor in range(1, 4):
    print(f"Floor {floor}:")
    booked = 0
    available = 0

    for room in range(1, 5):
        while True:
            answer = input(f"Room {room}: ").upper()
            if answer in ["Y", "N"]:
                break
            print("Sirf Y ya N likho.")

        if answer == "Y":
            booked += 1
        else:
            available += 1

    print(f"Floor {floor} = {booked} booked, {available} available")
    total_booked += booked
    total_available += available

print("\n===== SUMMARY =====")
print("Total Booked:", total_booked)
print("Total Available:", total_available)
```

## Invoice Tax Calculator
### Goal:

Take multiple items in multiple invoices and calculate tax per invoice.

### Steps:

Outer loop → for 3 invoices.

Inner loop → 4 items each.

Input item price, add to list.

Calculate total + 10% tax.

Print invoice total.

### Example:
    Invoice 1:
    Items: [100, 200, 300, 400]
    Invoice 2:
    Items: [250, 150, 350, 250]
    → Totals: [1100, 1100]

```python
invoice_totals = []

for invoice in range(1, 4):
    print(f"Invoice {invoice}:")
    items = []

    for item in range(1, 5):
        while True:
            try:
                price = float(input(f"Item {item} price: "))
                if price < 0:
                    print("Price negative nahi ho sakti.")
                    continue
                break
            except ValueError:
                print("Sirf number likho.")
        items.append(price)

    subtotal = sum(items)
    tax = subtotal * 0.10
    total = subtotal + tax

    invoice_totals.append(total)

    print("Items:", items)
    print(f"Subtotal: {subtotal}")
    print(f"Tax (10%): {tax}")
    print(f"Invoice Total: {total}\n")

print("Totals:", invoice_totals)
print("Grand Total:", sum(invoice_totals))
```

```python
for student in range(1, 6):
    print(f"\nStudent {student}:")
    marks = []
    failed_subjects = []

    for subject in range(1, 5):
        while True:
            try:
                m = int(input(f"Subject {subject} marks: "))
                if 0 <= m <= 100:
                    break
                print("Marks 0 se 100 ke beech likho.")
            except ValueError:
                print("Sirf number likho.")

        marks.append(m)

        if m < 33:
            print(f"Failed in Subject {subject}")
            failed_subjects.append(subject)

    total = sum(marks)
    percentage = total / 4

    if percentage >= 80:
        grade = "A"
    elif percentage >= 60:
        grade = "B"
    elif percentage >= 40:
        grade = "C"
    else:
        grade = "D"

    print(f"Student {student}: {marks}")
    print(f"Total: {total}/400, Percentage: {percentage:.2f}%")

    if failed_subjects:
        print("Result: FAIL (Failed in Subject", failed_subjects, ")")
    else:
        print("Grade:", grade)
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
features = ["Hotel", "Transport", "Food"]
averages = []

for package in range(1, 4):
    print(f"\nPackage {package}:")
    ratings = []

    for feature in features:
        while True:
            try:
                rating = int(input(f"{feature} rating (0-5): "))
                if 0 <= rating <= 5:
                    break
                print("Rating 0 se 5 ke beech likho.")
            except ValueError:
                print("Sirf number likho.")

        if rating == 0:
            print("Skipped")
            continue

        ratings.append(rating)

    if len(ratings) > 0:
        avg = sum(ratings) / len(ratings)
    else:
        avg = 0

    averages.append(avg)
    print(f"Package {package}: {ratings} -> Avg = {avg:.2f}")

print("\nAverages:", [round(a, 2) for a in averages])
best = averages.index(max(averages)) + 1
print("Best Package: Package", best)
```
