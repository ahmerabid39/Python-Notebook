```python
baggage_weight = float(input("Enter baggage weight (kg): "))
passport_valid = input("Is the passport valid? (Yes/No): ").strip().lower()
visa_status = input("Visa status (Valid/Invalid): ").strip().lower()
criminal_check = input("Criminal check (Clear/Flagged): ").strip().lower()

if passport_valid != "yes":
    print("Rejected: Invalid passport")
elif visa_status != "valid":
    print("Rejected: Invalid visa")
elif criminal_check == "flagged":
    print("Security Hold")
else:
    if baggage_weight > 30:
        print("Extra baggage charge applies")
    print("Cleared for boarding")
```

## Write a program to decide insurance eligibility.
Take input:
* age
* smoker (Yes/No)
* existing_disease (Yes/No)
* annual_income
#### Rules:
* age < 18 → reject
* smoker + disease → high premium category
* income < 200000 → limited coverage
* age > 60 → special medical check required
* otherwise → normal approval

```python
age = int(input("Enter your age: "))
smoker = input("Are you a smoker? (Yes/No): ").strip().lower()
existing_disease = input("Any existing disease? (Yes/No): ").strip().lower()
annual_income = float(input("Enter your annual income: "))

if age < 18:
    print("Rejected: Applicant must be at least 18")
else:
    results = []

    if smoker == "yes" and existing_disease == "yes":
        results.append("High premium category")
    if annual_income < 200000:
        results.append("Limited coverage")
    if age > 60:
        results.append("Special medical check required")

    if results:
        for r in results:
            print(r)
    else:
        print("Normal approval")
```

## Write a program to detect suspicious banking transactions.
Take input:
* transaction_amount
* location_match (Yes/No)
* time_of_transaction(Day/Night)
* previous_transactions_count
* device_verified (Yes/No)
#### Rules:
* amount > 100,000 → suspicious
* location mismatch + high amount → fraud alert
* night transaction + unverified device → high risk
* many transactions (>10 in 1 hour) → freeze account
* otherwise → normal transaction

```python
transaction_amount = float(input("Enter transaction amount: "))
location_match = input("Does the location match? (Yes/No): ").strip().lower()
time_of_transaction = input("Time of transaction (Day/Night): ").strip().lower()
previous_transactions_count = int(input("Number of transactions in the last hour: "))
device_verified = input("Is the device verified? (Yes/No): ").strip().lower()

alerts = []

if previous_transactions_count > 10:
    alerts.append("Freeze account")

if transaction_amount > 100000:
    alerts.append("Suspicious transaction")
    if location_match == "no":
        alerts.append("Fraud alert")

if time_of_transaction == "night" and device_verified == "no":
    alerts.append("High risk")

if alerts:
    for a in alerts:
        print(a)
else:
    print("Normal transaction")
```

## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
criminal_record = input("Criminal record? (Yes/No): ").strip().lower()
number_of_cases = int(input("Enter number of cases: "))
travel_history_flagged = input("Travel history flagged? (Yes/No): ").strip().lower()
financial_activity_score = float(input("Enter financial activity score (0-100): "))
last_activity_days = int(input("Days since last activity: "))

SUSPICIOUS_SCORE = 70  # scores above this count as suspicious

if criminal_record == "yes" and number_of_cases > 3:
    print("High Risk")
elif travel_history_flagged == "yes" and financial_activity_score > SUSPICIOUS_SCORE:
    print("High Risk")
elif last_activity_days > 365:
    print("Monitor")
elif (criminal_record == "no" and travel_history_flagged == "no"
      and financial_activity_score <= SUSPICIOUS_SCORE):
    print("Low Risk")
else:
    print("Medium Risk")
```
