### 20- Heat Transfer:

```python
m = int(input("what is the mass? "))
```

```python
c = int(input("what is the value of specific heat? "))
```

```python
t1 = int(input("what is the initial temperature? "))
```

```python
t2 = int(input("what is the final temperature? "))
```

```python
Q = m * c * (t2 - t1)
print(Q)
```
